```python
# import required libraries
import os
import json
import csv
```

```python
############## Create Objects
### Create an object for Contact's information

class Contact:
    def __init__(self,name,lastname,number):

        self.name=name
        self.lastname=lastname
        self.number=number


    def export_json(self):
        """
            Format the contact's information for export in JSON format
        """
        return {"first_name":self.name,
                "last_name":self.lastname,
                "phone_number":self.number}

    def export_csv(self):
        """
            Format the contact's information for export in CSV format
        """
        return [self.name,self.lastname,self.number]

    @staticmethod
    def import_json(jsonfile):
        """
            Load contact information from a json file item
        """
        return Contact(jsonfile["first_name"],
                       jsonfile["last_name"],
                       jsonfile["phone_number"])

    def import_csv(row):
        """
            Load contact information from a csv file row
        """
        return Contact(row[0],row[1],row[2])

    def __repr__(self):
        """
           Print the contact information
        """
        return f"{self.name} {self.lastname}, {self.number}"



# Create an object for Address book
class AddressBook:
    def __init__(self):
        self.contacts = []

    def add_contact(self):
        """
           Adds a new contact to the AddressBook,
           ensuring no contact with the same name and last name already exists
        """
        while True:
            new_contact = add_contact_info()
            contact_exists = False
            for existingcontact in self.contacts:
                if existingcontact.name == new_contact.name and existingcontact.lastname == new_contact.lastname:
                    contact_exists = True
                    break
            if contact_exists:
                print("A contact with this first and last name already exists. Please, modify your input.")
            else:
                self.contacts.append(new_contact)
                print("New contact added!")
                print("-------------------")
                break

    def show_existingContacts(self):
        """
           Print all contacts in AddressBook
        """
        if not self.contacts:
            print("No contact found!")
            return
        else:
            for i, contact in enumerate(self.contacts, 1):
                print(f"{i}. {contact}")
            print("----------------")


    def update_contact(self, name, lastname):
        """
           Allows to update the information of a given contact if found
           Args:
            name: First Name of the contact to update
            lastname: Last Name of the contact to update
        """

        matchcontacts = [
            (i, c) for i, c in enumerate(self.contacts)
            if name in c.name and lastname in c.lastname
        ]


        if len(matchcontacts) == 0:
            print("No contact found with such name")
            return


        index, contact = matchcontacts[0]

        print(f"Contact found: {contact.name} {contact.lastname} with the following number  {contact.number}")

        print("Please, enter the new information")
        new_name=input("Please, enter the new name: ")
        new_lastname=input("Please, enter the new last name: ")
        new_number = input("Please, enter the new number: ")
        while not new_number.isdigit():
            new_number = input("Please, enter a valid phone number: ")

        contact.name = new_name
        contact.lastname = new_lastname
        contact.number = new_number


    def delete_contact(self, name, lastname):
        """
           Allows to delete the information of a given contact

           Args:
            name: First Name of the contact to delete
            lastname: Last Name of the contact to delete
        """

        matchcontacts = [
            (i, c) for i, c in enumerate(self.contacts)
            if name in c.name and lastname in c.lastname
        ]
        if len(matchcontacts) == 0:
            print("No contact found with such name")
            return

        index, contact = matchcontacts[0]
        deleted = self.contacts.pop(index)
        print(f"Contact removed: {deleted.name} {deleted.lastname}")



    def find_existing_contact(self, name, lastname):
        """
            Search for a contact by name and last name and
            print the contact's information if found.
        Args:
            filename (_type_): _description_
            format (_type_): _description_
        """

        results = [
            c for c in self.contacts
            if name in c.name or lastname in c.lastname
        ]
        if len(results)==0:
            print("No contact found")
            print("------")
        else:

            for contact in results:
                print(f" {contact}")
                print("----------------")



    def save_file(self,filename, format):
        """
            Saves AddressBook's contacts in a file, JSON or CSV.
        Args:
            filename : The name of the file where contacts will be saved
            format:  The format in which to save the contacts. Accepts 'json' or 'csv'
        """
        if format=="json":
            with open(filename, "w") as f:
                json.dump([c.export_json() for c in self.contacts], f)
        elif format=="csv":
            with open(filename, mode="w", newline='', encoding="utf-8") as f:
                writer = csv.writer(f)
                for contact in self.contacts:
                    writer.writerow(contact.export_csv())



    def load_file(self, filename,format):

        """
            Load AddressBook's contacts from an existing file, JSON or CSV.
        Args:
            filename : The name of the file where contacts are saved
            format:  The format in which the contacts are saved. Accepts 'json' or 'csv'
        """

        if format=="json":
            with open(filename, "r") as f:
                data = json.load(f)
                self.contacts = [Contact.import_json(d) for d in data]
        if format=="csv":
            with open(filename, mode="r", encoding="utf-8") as f:
                reader = csv.reader(f)

                self.contacts = [Contact.import_csv(row) for row in reader]

```

```python
### Define functions
#
def show_menu_option():

    """
      Shows options and receives user's input
    """

    print("1. Add New Contact")
    print("2. Show Existing Contacts")
    print("3. Update Contact")
    print("4. Delete Contact")
    print("5. Find Contact")
    print("6. Save Contacts")
    print("7. Load Contacts")
    print("8. Show menu options")
    print("9. Exit")


def inputuser():
    while True:

        userinput=input("Please, select an option: ")
        if userinput.isdigit():
            break
        else:

            print("Invalid input. Numbers only, please select a valid option: ")
    return int(userinput)




def add_contact_info():
    """
        Function required for enter new contact's information
    """
    name=input("First Name: ")
    lastname=input("Last Name: ")
    while True:
        number=input("Phone Number: ")
        if number.isdigit():
            break
        else:
            print("Please, enter a valid phone number")
    return Contact(name,lastname,number)


def load_addressbook(Addressbook):

    filename=input("Please, provide the name of the file to load: ")
    json_file = filename+".json"
    csv_file = filename+".csv"
    json_exist = os.path.exists(json_file)
    csv_exist = os.path.exists(csv_file)

    if json_exist and csv_exist:
        print("Found both JSON and CVS files. Which one do you want to load?")
        print("1. JSON")
        print("2. CSV")
        print("3. Skip")


        format = int(input("Plese,select a file format: "))

        if format==1:
            Addressbook.load_file(json_file,"json")
        elif format == 2:
            Addressbook.load_file(csv_file,"csv")
        elif format==3:
            return

    elif json_exist:
        scelta = input("Found JSON file. Do you want to load it? (Yes/No) ")
        if scelta == "Yes":
            Addressbook.load_file(json_file,"json")
    elif csv_exist:
        scelta = input("Found CVS file. Do you want to load it? (Yes/No)")
        if scelta == "Yes":
            Addressbook.load_file(csv_file,"csv")
    else:
        print("No file found with such name")
```

```python
def main():
    Addressbook = AddressBook()

    print("\n-------- Welcome to Contact Ease Solutions AddressBook --------")
    print("\n-------- How can I help you? --------")

    show_menu_option()

    while True:
        userchoice=inputuser()
        if userchoice==1:
            print("---- Add New Contact ----")
            print("Please, provide the following information:")
            Addressbook.add_contact()
        elif userchoice==2:
            print("---- Show Existing Contacts ----")
            Addressbook.show_existingContacts()

        elif userchoice==3:
            print("---- Update Contact ----")
            print("Please, provide the information for the contact you want to update")
            name=input("First Name: ")
            lastname=input("Last Name: ")
            Addressbook.update_contact(name,lastname)

        elif userchoice==4:
            print("---- Delete Contact ----")
            print("Please, provide the information for the contact to delete")
            name=input("First Name: ")
            lastname=input("Last Name: ")
            Addressbook.delete_contact(name,lastname)


        elif userchoice==5:
            print("---- Find Contact ----")
            name=input("First Name: ")
            lastname=input("Last Name: ")
            Addressbook.find_existing_contact(name,lastname)


        elif userchoice==6:
            print("---- Save Contacts ----")

            filename=input("Please, provide the name to save the: ")
            print("File Formats available: ")
            print("1. JSON")
            print("2. CSV")
            formatuser = int(input("Plese,select a file format: "))

            if formatuser==1:
                format="json"
            elif formatuser == 2:
                format="csv"
            Addressbook.save_file(filename+"."+format,format)

        elif userchoice==7:
            print("---- Load Contacts ----")
            load_addressbook(Addressbook)

        elif userchoice==8:
            print("---- Show menu options ----")
            show_menu_option()
        elif userchoice==9:
            print("---- Exit ----")
            print("Session ended. Thank you for using ConcatEase Solutions.")
            print("We hope to see you again soon!")
            exit()

        else:
            print("Please, select a valid option")
            show_menu_option()



if __name__ =="__main__":

    main()
```
